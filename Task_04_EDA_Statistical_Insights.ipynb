```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("clean_dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
numeric_cols = ["Sales", "Quantity", "Discount", "Profit"]

statistics = pd.DataFrame({
    "Mean": df[numeric_cols].mean(),
    "Median": df[numeric_cols].median(),
    "Standard Deviation": df[numeric_cols].std(),
    "Q1": df[numeric_cols].quantile(0.25),
    "Q3": df[numeric_cols].quantile(0.75)
})

statistics
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Sales"], bins=30, kde=True)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Profit"], bins=30, kde=True)

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Discount"], bins=20, kde=True)

plt.title("Discount Distribution")
plt.xlabel("Discount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Sales"])

plt.title("Sales Box Plot")
plt.xlabel("Sales")

plt.show()
```

```python
correlation_matrix = df[numeric_cols].corr()

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    hue="Discount"
)

plt.title("Sales vs Profit by Discount")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Profit"
)

plt.title("Quantity vs Profit")
plt.xlabel("Quantity")
plt.ylabel("Profit")

plt.show()
```

```python
discount_profit = df[["Discount", "Profit"]].corr()

print("Hypothesis 1:")
print("Higher discounts are associated with lower profit.")

print("\nCorrelation between Discount and Profit:")
print(round(discount_profit.loc["Discount", "Profit"], 3))
```

```python
sales_profit = df[["Sales", "Profit"]].corr()

print("Hypothesis 2:")
print("Higher sales are associated with higher profit.")

print("\nCorrelation between Sales and Profit:")
print(round(sales_profit.loc["Sales", "Profit"], 3))
```

```python
quantity_profit = df[["Quantity", "Profit"]].corr()

print("Hypothesis 3:")
print("Order quantity is associated with profit.")

print("\nCorrelation between Quantity and Profit:")
print(round(quantity_profit.loc["Quantity", "Profit"], 3))
```

```python
discount_analysis = df.groupby("Discount")["Profit"].mean().sort_values(ascending=False)

discount_analysis
```

```python
category_analysis = df.groupby("Category")[["Sales", "Profit"]].mean()

category_analysis
```

```python
category_analysis.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.title("Average Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Average Value")

plt.xticks(rotation=0)

plt.show()
```

## Top 5 Critical Findings

1. **Sales and Profit Relationship:** Sales and Profit show a moderate positive correlation of 0.45, indicating that higher sales values are generally associated with higher profit.

2. **Discount and Profit Relationship:** Discount and Profit show a negative correlation of -0.51, indicating that higher discount levels are associated with lower profit in the dataset.

3. **Quantity and Profit Relationship:** Quantity and Profit have a weak positive correlation of 0.20, suggesting that order quantity has a limited positive relationship with profit.

4. **Sales and Quantity Relationship:** Sales and Quantity show a weak positive correlation of 0.27, indicating that larger quantities are somewhat associated with higher sales.

5. **Discount and Sales Relationship:** Sales and Discount have an almost zero correlation of -0.01, suggesting that there is very little linear relationship between these two variables.

> Note: Correlation indicates association between variables and does not by itself establish a causal relationship.

## Business Hypothesis Conclusions

### Hypothesis 1
**Higher discounts are associated with lower profit.**

The correlation between Discount and Profit is **-0.51**, showing a moderate negative relationship. Therefore, the dataset shows an association between higher discount levels and lower profit.

### Hypothesis 2
**Higher sales are associated with higher profit.**

The correlation between Sales and Profit is **0.45**, showing a moderate positive relationship. Therefore, higher sales values are generally associated with higher profit.

### Hypothesis 3
**Order quantity is associated with profit.**

The correlation between Quantity and Profit is **0.20**, showing a weak positive relationship. Therefore, quantity has a limited positive association with profit.
